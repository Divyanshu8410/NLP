```python

# Install required libraries:
# pip install scikit-learn

import os
import shutil
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

# 1. Sample Training Documents
documents = [
    "Artificial intelligence and machine learning technology",
    "Python programming and software development",
    "Computer networks and data science",
    "Cricket football and sports tournament",
    "Indian cricket team won the match",
    "Football players scored goals",
    "Government election and political leaders",
    "Parliament and political discussions",
    "Prime minister announced new policies",
    "Stock market and business investment",
    "Company profits and financial growth",
    "Banking and business management"
]

# 2. Document Categories
categories = [
    "Technology", "Technology", "Technology",
    "Sports", "Sports", "Sports",
    "Politics", "Politics", "Politics",
    "Business", "Business", "Business"
]

# 3. Create ML Classification Model
model = make_pipeline(
    TfidfVectorizer(),
    MultinomialNB()
)

# 4. Train Model
print("Training Document Classification Model...")
model.fit(documents, categories)

# 5. New Documents for Classification
new_documents = [
    "Python is used for artificial intelligence",
    "India won the cricket world cup",
    "Government announced new election policies",
    "Company shares increased in the stock market"
]

# 6. Create Main Output Folder
output_folder = "classified_documents"
os.makedirs(output_folder, exist_ok=True)

# 7. Classify and Separate Documents
print("\nDocument Classification Results:\n")

predictions = model.predict(new_documents)

for i, (doc, category) in enumerate(zip(new_documents, predictions)):

    # Create category folder
    category_folder = os.path.join(output_folder, category)
    os.makedirs(category_folder, exist_ok=True)

    # Save document in its category folder
    file_path = os.path.join(
        category_folder,
        f"document_{i+1}.txt"
    )

    with open(file_path, "w", encoding="utf-8") as file:
        file.write(doc)

    print(f"Document: {doc}")
    print(f"Category: {category}")
    print("-" * 50)

# 8. Display Final Result
print("\nDocuments separated successfully!")

for category in os.listdir(output_folder):
    print(f"{category}:")
    print(os.listdir(os.path.join(output_folder, category)))

```
