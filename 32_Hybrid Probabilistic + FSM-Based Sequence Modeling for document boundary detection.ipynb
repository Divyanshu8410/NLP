```python

import re
import numpy as np

# 1. Sample Input Document
text = """
INVOICE
Invoice Number: 101
Date: 12-10-2026
Total Amount: 5000

RECEIPT
Receipt Number: 202
Payment Received: 5000

REPORT
Title: Annual Sales Report
Revenue increased by 20 percent.

INVOICE
Invoice Number: 102
Date: 15-10-2026
Total Amount: 8000
"""

# 2. Split Text into Lines
lines = [line.strip() for line in text.split("\n") if line.strip()]

# 3. Define Document Keywords
document_keywords = [
    "INVOICE",
    "RECEIPT",
    "REPORT"
]

# 4. Probabilistic Boundary Detection
def calculate_probability(line):

    score = 0

    if line.upper() in document_keywords:
        score += 0.6

    if re.search(r"invoice|receipt|report", line, re.I):
        score += 0.2

    if line.isupper():
        score += 0.2

    return min(score, 1.0)

# 5. Define FSM States
NORMAL = "NORMAL"
NEW_DOCUMENT = "NEW_DOCUMENT"

state = NORMAL

documents = []
current_document = []

# 6. Hybrid Probabilistic + FSM Model
for line in lines:

    probability = calculate_probability(line)

    # Check document boundary
    if probability >= 0.6:
        if state == NORMAL:

            if current_document:
                documents.append(current_document)
                current_document = []

            state = NEW_DOCUMENT

    else:
        state = NORMAL

    current_document.append(line)

# Add Last Document
if current_document:
    documents.append(current_document)

# 7. Display Results
print("Document Boundary Detection Results")
print("------------------------------------")

for i, document in enumerate(documents, 1):

    print(f"\nDocument {i}:")
    for line in document:
        print(line)

print("\nTotal Documents Detected:", len(documents))
```
