```python
import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk.chunk import RegexpParser

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)

nlp = spacy.load("en_core_web_sm")

print("Parsing and Chunking setup completed!")
```

```python
sentence = "Dev is developing an intelligent chatbot using Python."

# ---------- RegEx Chunking ----------
print("========== REGEX CHUNKING ==========")

words = word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)

# Grammar for Noun Phrase (NP)
grammar = r"""
    NP: {<DT>?<JJ>*<NN.*>+}
"""

chunk_parser = RegexpParser(grammar)
chunk_tree = chunk_parser.parse(pos_tags)

print("\nPOS Tagged Sentence:")
print(pos_tags)

print("\nChunking Result:")
chunk_tree.pretty_print()


# ---------- spaCy Parsing ----------
print("\n========== SPACY DEPENDENCY PARSING ==========")

doc = nlp(sentence)

print("\nWord\tPOS\tDependency\tHead")
print("-" * 45)

for token in doc:
    print(f"{token.text}\t{token.pos_}\t{token.dep_}\t\t{token.head.text}")
```
