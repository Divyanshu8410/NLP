```python
import pandas as pd
from textblob import TextBlob

# Step 1: Create a product/service reviews dataset
data = {
    "Review": [
        "The product is excellent and very useful",
        "Amazing quality and fast delivery",
        "I love this product",
        "The service was very good",
        "The product is okay",
        "Average experience, nothing special",
        "The product is terrible",
        "Very poor quality and bad service",
        "I hate this product",
        "Delivery was slow and disappointing"
    ]
}

df = pd.DataFrame(data)

# Step 2: Perform opinion mining
def analyze_sentiment(review):
    polarity = TextBlob(review).sentiment.polarity

    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    return pd.Series([polarity, sentiment])

df[["Polarity", "Opinion"]] = df["Review"].apply(analyze_sentiment)

# Step 3: Display results
print("OPINION MINING RESULTS\n")
print(df.to_string(index=False))

# Step 4: Count positive, negative and neutral opinions
print("\nOPINION SUMMARY")
print(df["Opinion"].value_counts())

# Step 5: Calculate opinion percentages
print("\nOPINION PERCENTAGES")
print((df["Opinion"].value_counts(normalize=True) * 100).round(2))
```
