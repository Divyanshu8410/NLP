```python
import re
import pandas as pd
import spacy

# Load NLP model
try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    raise SystemExit(
        "Run: !pip install spacy pandas && "
        "python -m spacy download en_core_web_sm"
    )

# 1. Structured document (CSV-like data)
structured_data = {
    "Name": ["Rahul Sharma", "Priya Verma"],
    "Email": ["rahul@gmail.com", "priya@yahoo.com"],
    "Salary": [50000, 60000]
}

df = pd.DataFrame(structured_data)

print("STRUCTURED DOCUMENT - Extracted Data")
print(df.to_string(index=False))

# 2. Unstructured document (free text)
text = """
Rahul Sharma works at Google in Delhi.
Contact him at rahul@gmail.com.
Priya Verma joined Microsoft in Bangalore.
Her email is priya@yahoo.com.
"""

doc = nlp(text)

# Extract named entities
print("\nUNSTRUCTURED DOCUMENT - Named Entities")
for entity in doc.ents:
    print(entity.text, "->", entity.label_)

# Extract email addresses
emails = re.findall(
    r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b',
    text
)

print("\nExtracted Emails:")
for email in emails:
    print(email)

# Extract organizations and locations
organizations = [ent.text for ent in doc.ents if ent.label_ == "ORG"]
locations = [ent.text for ent in doc.ents if ent.label_ == "GPE"]

print("\nOrganizations:", organizations)
print("Locations:", locations)

# Save structured data to CSV
df.to_csv("extracted_data.csv", index=False)
print("\nStructured data saved to extracted_data.csv")
```
