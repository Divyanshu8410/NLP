## Cosine Similarity Computation for Text Documents

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Define sample documents
documents = [
    "The quick brown fox jumps over the lazy dog",
    "Never jump over the lazy dog often",
    "The dog is very lazy",
    "The fox is quick and brown"
]

print("Original Documents:")
for i, doc in enumerate(documents):
    print(f"Document {i+1}: {doc}")
```

### 2. Text Vectorization using TF-IDF

TF-IDF (Term Frequency-Inverse Document Frequency) is a numerical statistic that is intended to reflect how important a word is to a document in a collection or corpus. We'll use `TfidfVectorizer` from `scikit-learn` to convert our text documents into a matrix of TF-IDF features.

```python
# Initialize TF-IDF Vectorizer
# stop_words='english' removes common English words that don't add much meaning.
# lowercase=True is the default, ensuring consistent casing.
vectorizer = TfidfVectorizer(stop_words='english')

# Fit the vectorizer to the documents and transform the documents into TF-IDF vectors
tfidf_matrix = vectorizer.fit_transform(documents)

# Get the feature names (words) to understand the vectors
feature_names = vectorizer.get_feature_names_out()

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
print("Vocabulary (Feature Names):")
print(feature_names[:10]) # Display first 10 for brevity

# You can view the TF-IDF matrix (sparse format) or convert to dense for inspection
# For larger datasets, keep it sparse for memory efficiency.
# print("TF-IDF Matrix (dense, first 2 documents):")
# display(pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names).head(2))
```

### 3. Compute Cosine Similarity

Cosine similarity measures the cosine of the angle between two non-zero vectors. A cosine similarity of 1 means the vectors are identical, 0 means they are orthogonal (no similarity), and -1 means they are exactly opposite.

We will compute the cosine similarity between all pairs of documents.

```python
# Compute cosine similarity between all document pairs
cosine_sim_matrix = cosine_similarity(tfidf_matrix, tfidf_matrix)

print("Cosine Similarity Matrix:")
# Display the matrix with document labels for better readability
import pandas as pd
sim_df = pd.DataFrame(cosine_sim_matrix,
                      index=[f"Doc {i+1}" for i in range(len(documents))],
                      columns=[f"Doc {i+1}" for i in range(len(documents))])

display(sim_df)

print("\nInterpretation:")
print("- Values closer to 1 indicate higher similarity.")
print("- Values closer to 0 indicate lower similarity.")
print("For example, Document 1 and Document 4 (fox, quick, brown) have a higher similarity than Document 1 and Document 3 (quick, brown, lazy dog vs lazy dog).")
```
