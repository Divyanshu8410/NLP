```python
import gensim.downloader as api

# Load GloVe embeddings
print("Loading GloVe embeddings...")
glove = api.load("glove-wiki-gigaword-50")
print("Embeddings loaded successfully!\n")

# Words to represent as vectors
words = ["king", "queen", "man", "woman"]

# Display vector representations
for word in words:
    print(f"Word: {word}")
    print("Vector:", glove[word])
    print("Dimensions:", len(glove[word]))
    print()

# Calculate similarity
print("Similarity between king and queen:",
      glove.similarity("king", "queen"))
```
