```python

# Install: pip install transformers torch

from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# 1. Load Pre-trained DialoGPT Model
print("Loading Chatbot...")

tokenizer = AutoTokenizer.from_pretrained("microsoft/DialoGPT-small")
model = AutoModelForCausalLM.from_pretrained("microsoft/DialoGPT-small")

# 2. Start Chatbot
print("\nChatbot is Ready!")
print("Type 'exit' to stop.\n")

chat_history = None

# 3. Chat Loop
while True:

    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Bot: Goodbye!")
        break

    # 4. Convert Input into Tokens
    new_input = tokenizer.encode(
        user_input + tokenizer.eos_token,
        return_tensors="pt"
    )

    # 5. Add Previous Conversation
    if chat_history is not None:
        bot_input = torch.cat([chat_history, new_input], dim=-1)
    else:
        bot_input = new_input

    # 6. Generate Response
    chat_history = model.generate(
        bot_input,
        max_new_tokens=50,
        pad_token_id=tokenizer.eos_token_id,
        do_sample=True,
        temperature=0.7,
        top_k=50
    )

    # 7. Display Response
    response = tokenizer.decode(
        chat_history[:, bot_input.shape[-1]:][0],
        skip_special_tokens=True
    )

    print("Bot:", response)

```
