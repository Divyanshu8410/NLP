```python
import gensim.downloader as api
from gensim.similarities import WmdSimilarity

# Step 1: Load pre-trained word embeddings
print("Loading word embeddings...")
model = api.load("glove-wiki-gigaword-50")
print("Embeddings loaded successfully!\n")

# Step 2: Input two texts
text1 = "The king rules the kingdom"
text2 = "The queen rules the country"

# Step 3: Preprocess the texts
words1 = text1.lower().split()
words2 = text2.lower().split()

# Step 4: Calculate Word Mover's Distance
distance = model.wmdistance(words1, words2)

# Step 5: Display results
print("Text 1:", text1)
print("Text 2:", text2)
print("Word Mover's Distance:", distance)

# Step 6: Convert distance into a similarity score
similarity = 1 / (1 + distance)
print("Similarity Score:", similarity)
```
