```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report

# Step 1: Sample dataset
texts = [
    "I love this movie",
    "This film is amazing",
    "The movie was fantastic",
    "I hate this movie",
    "This film is terrible",
    "The movie was boring",
    "The product is excellent",
    "I really like this product",
    "This product is awful",
    "I dislike this product",
    "The food was delicious",
    "The food was disgusting"
]

labels = [
    "Positive", "Positive", "Positive",
    "Negative", "Negative", "Negative",
    "Positive", "Positive", "Negative", "Negative",
    "Positive", "Negative"
]

# Step 2: Convert text into TF-IDF vectors
tfidf = TfidfVectorizer()
X = tfidf.fit_transform(texts)
y = labels

# Step 3: Split dataset into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# Step 4: Train Naive Bayes model
nb = MultinomialNB()
nb.fit(X_train, y_train)
nb_pred = nb.predict(X_test)

# Step 5: Train SVM model
svm = LinearSVC()
svm.fit(X_train, y_train)
svm_pred = svm.predict(X_test)

# Step 6: Evaluate both models
print("NAIVE BAYES RESULTS")
print("Accuracy:", accuracy_score(y_test, nb_pred))
print(classification_report(y_test, nb_pred, zero_division=0))

print("\nSVM RESULTS")
print("Accuracy:", accuracy_score(y_test, svm_pred))
print(classification_report(y_test, svm_pred, zero_division=0))

# Step 7: Predict a new text
new_text = ["This movie is amazing and fantastic"]
new_vector = tfidf.transform(new_text)

print("Naive Bayes Prediction:", nb.predict(new_vector)[0])
print("SVM Prediction:", svm.predict(new_vector)[0])
```
