```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

stop_words = set(stopwords.words('english'))

print("Stop-word removal setup completed!")
```

```python
text = """
Natural Language Processing is a branch of Artificial Intelligence.
It allows computers to understand and process human language.
NLP is widely used in chatbots, translation, and text analysis.
"""

# Tokenization
words = word_tokenize(text)

# Remove stopwords and punctuation
filtered_words = [
    word for word in words
    if word.lower() not in stop_words and word.isalpha()
]

print("========== ORIGINAL TEXT ==========")
print(text)

print("========== WORDS BEFORE REMOVAL ==========")
print(words)

print("\n========== WORDS AFTER STOP-WORD REMOVAL ==========")
print(filtered_words)
```
