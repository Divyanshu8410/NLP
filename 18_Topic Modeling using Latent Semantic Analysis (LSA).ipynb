```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

# Step 1: Sample documents
documents = [
    "Machine learning and artificial intelligence are transforming technology",
    "Deep learning uses neural networks for artificial intelligence",
    "Python programming is useful for data science and machine learning",
    "Football and cricket are popular sports around the world",
    "The football team won the match with excellent performance",
    "Cricket is a popular sport played by many teams"
]

# Step 2: Text preprocessing
def preprocess(text):
    text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    return text

clean_documents = [preprocess(doc) for doc in documents]

# Step 3: Convert text into TF-IDF matrix
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(clean_documents)

# Step 4: Apply LSA using Truncated SVD
lsa = TruncatedSVD(n_components=2, random_state=42)
document_topics = lsa.fit_transform(tfidf_matrix)

# Step 5: Display topics and important words
terms = vectorizer.get_feature_names_out()

print("Topics discovered using LSA:\n")

for topic_idx, topic in enumerate(lsa.components_):
    top_indices = topic.argsort()[-5:][::-1]
    top_words = [terms[i] for i in top_indices]
    print(f"Topic {topic_idx + 1}: {', '.join(top_words)}")

# Step 6: Display topic representation of each document
print("\nDocument Topic Representation:\n")

for i, topic_values in enumerate(document_topics):
    dominant_topic = topic_values.argmax() + 1
    print(f"Document {i + 1}: Topic {dominant_topic}")
    print("Topic scores:", topic_values.round(3))
```
