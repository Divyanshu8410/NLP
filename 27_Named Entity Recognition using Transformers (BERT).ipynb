```python

# Install required library:
# pip install transformers torch

from transformers import pipeline

# 1. Load Pre-trained BERT NER Model
print("Loading BERT NER Model...")

ner = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)

# 2. Input Text
text = """
Elon Musk is the CEO of Tesla.
Google is located in California.
Virat Kohli is a famous Indian cricketer.
"""

print("\nInput Text:")
print(text)

# 3. Perform Named Entity Recognition
results = ner(text)

# 4. Display Extracted Entities
print("\nNamed Entities:")

for entity in results:
    print(
        "Entity:", entity["word"],
        "| Type:", entity["entity_group"],
        "| Confidence:", round(entity["score"], 3)
    )

print("\nNER Completed Successfully!")
```
