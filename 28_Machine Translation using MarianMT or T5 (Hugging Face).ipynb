```python

# Install required library:
# pip install transformers sentencepiece torch

from transformers import MarianMTModel, MarianTokenizer

# 1. Select Translation Model (English to Hindi)
model_name = "Helsinki-NLP/opus-mt-en-hi"

# 2. Load Tokenizer and Model
print("Loading Translation Model...")

tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

# 3. Input English Text
text = "Artificial Intelligence is changing the world."

print("\nOriginal English Text:")
print(text)

# 4. Tokenize Input Text
tokens = tokenizer(
    text,
    return_tensors="pt",
    padding=True,
    truncation=True
)

# 5. Generate Translation
translated = model.generate(**tokens)

# 6. Decode Translated Text
translation = tokenizer.decode(
    translated[0],
    skip_special_tokens=True
)

# 7. Display Result
print("\nTranslated Hindi Text:")
print(translation)

print("\nTranslation Completed Successfully!")
```
