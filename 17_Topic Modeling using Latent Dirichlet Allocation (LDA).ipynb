```python
import nltk
import re
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation

# Download stopwords
nltk.download('stopwords', quiet=True)
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

# Sample documents
documents = [
    "Machine learning and artificial intelligence are transforming technology.",
    "Deep learning uses neural networks for artificial intelligence.",
    "Football and cricket are popular sports around the world.",
    "The team won the football match with excellent performance.",
    "Python programming is useful for data science and machine learning.",
    "Data science uses Python for data analysis and visualization."
]

# Text preprocessing
def preprocess(text):
    text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    words = text.split()
    return ' '.join(word for word in words if word not in stop_words)

clean_documents = [preprocess(doc) for doc in documents]

# Convert text into a document-term matrix
vectorizer = CountVectorizer()
doc_term_matrix = vectorizer.fit_transform(clean_documents)

# Apply LDA topic modeling
lda = LatentDirichletAllocation(
    n_components=2,
    random_state=42,
    max_iter=10
)
lda.fit(doc_term_matrix)

# Display topics and their top words
feature_names = vectorizer.get_feature_names_out()

print("Topics discovered by LDA:\n")
for topic_idx, topic in enumerate(lda.components_):
    top_words = [
        feature_names[i]
        for i in topic.argsort()[:-6-1:-1]
    ]
    print(f"Topic {topic_idx + 1}: {', '.join(top_words)}")

# Display dominant topic for each document
print("\nDocument Topic Assignments:\n")
topic_distribution = lda.transform(doc_term_matrix)

for i, distribution in enumerate(topic_distribution):
    dominant_topic = distribution.argmax() + 1
    print(f"Document {i + 1}: Topic {dominant_topic}")
    print("Topic probabilities:", distribution.round(3))
```
