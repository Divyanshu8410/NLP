```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

Let's start by defining some example text documents.

```python
documents = [
    "The quick brown fox jumps over the lazy dog.",
    "Never jump over the lazy dog quickly.",
    "The dog is quick and brown."
]
display(documents)
```

Now, we'll create a `CountVectorizer` instance, fit it to our documents, and transform the documents into a bag-of-words representation.

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(documents)

# Display the vocabulary (the unique words found)
print("Vocabulary:")
display(vectorizer.get_feature_names_out())

# Display the bag-of-words representation as a DataFrame for better readability
bow_df = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())
display(bow_df)
```
